```python
from edsl import QuestionFreeText, QuestionMultipleChoice, Model, Survey, Agent, Scenario, FileStore
```

```python
model = Model("meta-llama/Llama-3.3-70B-Instruct", service_name="goodfire")
```

```python
q1 = QuestionMultipleChoice(
    question_name="q1",
    question_text="What is the capital of France?",
    question_options=["Paris", "London", "Berlin"],
)

response = q1.by(model).run(cache = False)
```
